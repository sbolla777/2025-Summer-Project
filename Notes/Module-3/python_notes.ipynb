```python
str = "  hello, world  ".title()
print(str)
str1 = "Say" +  str[:7]
# this will print "Say  Hello," the first five characters with "Say" concatenated
print(str1)
str2 = str[-7:-1]
# since we used backward indexing, this will print "World"
print(str2)
str = str.replace("World", "Everyone")
# due to the replace function, "  Hello, Everyone  " will be printed
print(str)
str = str.lower()
# the lowercase version "  hello, everyone  " will be printed
print(str)
str = str.strip()
# without whitespaces, "hello, everyone" will be printed
print(str)
split_str = str.split(",")
# the string has been split around the comma and stripped, so the second part "everyone" will be printed
print(split_str[1].strip())
```

# Datatypes

Python **lists** are used to store items in an ordered, indexed variable that can be modified. Like strings, lists have backward and forward indexing. However, instead of accessing characters, we can access whichever object was placed at the index. In order to find an item within the list, the method index(item) will return the first index where the item occurs. Or, the count(item) method will sum the number of occurences of the item in the list.

Lists can also be **sliced**, using a similar format to strings. To get the first five elements of a list, we would use list[:5]. The last four elements would be list[-5:]. The call list[3:7] would get the elements from the third position to the sixth.

In order to add an item to a list, we use the append(item) function. To put an element in a specific position, the insert(index, item) call is used. Conversely, the remove(item) function deletes the first occurence of the element. To empty the entire list, the clear() function can be utilized. To reverse the contents of the list, we use the reverse() function, and to sort the list we can use the sort() function. To join two different lists either the '+' operator or the extend(list2) function can be used.

Characteristics of the list can also be obtained using functions. The number of total elements in a list is found with the len(list) function. The maximum and minimum values are calculated with the max(list) and min(list) functions, respectively. For lists comprised of only ints or floats, the sum(list) function finds the sum of all values. To copy the list we can use copy(), and to create a sorted copy of the list we use sorted(list).

To loop through a list, our for loop must use "for item in list" as its header. Then, each item can be modified or used within the loop's body.

```python
# the list below is the grades of a class on a math test
grades = [85, 23, 90, 98, 100, 55, 67, 48, 67, 88, 92, 92, 45, 77, 75, 74, 75]
# the call below shows the number of grades, or the number of students
num = len(grades)
print(f"There are {num} students in the class")
# someone joined the class late and had to make up the exam
grades.append(95)
# the teacher gave fifteen points to the person with the lowest grade
lowest = min(grades)
grades.remove(lowest)
grades.append(lowest + 15)
# the teacher wishes to find the five highest scores
grades.sort()
highest = grades[-5:]
print(f"The five highest grades on the test were {highest}")
# the teacher wants to find the average score
avg = sum(grades)/num
print(f"The average score on the test was {avg}")
# the teacher wishes to award points based on how low a student scored, with 0 points for the highest and 16 for the lowest
grades.reverse()
bonus = 0
new_grades = []
for grade in grades:
  new_grades.append(grade + bonus)
  bonus += 1
# display the new grades
print(f"Here are the new grades: {new_grades}")
# another class is also reporting their grades
other_grades = [45, 43, 55, 66]
all_grades = new_grades + other_grades
# display the grades
print(f"Here are all the grades: {all_grades}")
```

Like lists, **sets** can have elements added and removed. Unlike tuples and lists, however, this data type is also unordered and does not allow for duplicate values.

To add an item to a set, we can use add(item). The call update(set2) adds a second set's elements to the original set. To remove an item that is defintely in the list the method remove(item) can be used. Otherwise, the call discard(item) will not raise any errors. pop() can be used to remove and obtain a random item from the set. When emptying the entire set, the clear() function can be called.

Several operations on sets from set theory are also built into Python such as **union**(other_sets), **intersection**(other_sets), and **difference**(other_sets). Each of these function produces a new set with the given operation performed between the first and second sets. The **symmetric_difference**(set2) method returns a new set which contains all elements in either set but not in intersection. To check whether a set is a **superset**, **subset**, or **disjoint** of another set, we use the issuperset(set2), issubset(set2), and isdisjoint(set2) methods.

Like with the previous data types, the len(set) function can be used to find the sumber of items in the set. max(set) and min(set) return the maximum and mininum of the set. Furthermore, sorted(set) returns a sorted version of the set.

Looping over tuples is also the same as for the other types, and our header will be: for item in set. We can perform methods using the item and other set operations in the body.

```python
# the students athletes at a highschool
tennis = {"Aiden", "Max", "Carly", "Lily", "Zara", "Ethan", "Frankie", "Rose", "Taylor", "Riley", "Avery"}
soccer = {"Cassie", "Sarah", "Jennie", "Jacob", "Timothy", "River", "Max", "Avery", "Lily", "Jackson"}
track = {"Jacob", "Max", "Lily", "Jackson", "Marie", "Megan", "Ben", "John", "Aaron", "Eric", "Christina", "April"}
# which students play both tennis and run track?
both = tennis.intersection(track)
print(f"These students play both tennis and track: {both}")
# which students play only one sport?
one = tennis.symmetric_difference(track.union(soccer))
print(f"These students play only one sport: {one}")
# how many student athletes are there?
athletes = tennis.union(soccer, track)
print(f"There are {len(athletes)} students athletes")
# three students are friends outside of school. do they all play tennis?
friends = {"Max", "Lily", "Taylor"}
if friends.issubset(tennis):
  print("Yes, they all play tennis")
# do any of the friends play soccer?
if not friends.isdisjoint(soccer):
  print("Yes, they do")
```

To create an empty dictionary, a copy, or convert another data structure into a dictionary, we can use dict(object). As always, the del keyword will delete any entire dictionary object.

The format for looping over dictionaries is slightly different. We can either loop directly over the dictionary, which is equivalent to looping over its keys or over dictionary.keys(). # Operators

**Arithmetic** operators are used to perform mathematical operation using numbers. This includes addition ('**+**'), subtraction ('**-**'), multiplication ('**\***'), and division ('**/**'). Less straightforward operators are the modulo operation ('**%**'), floor division ('**//**'), and exponents ('**\****'). The modulo operation finds the remainder of a division calculation. Conversely, floor division discards the remainder and returns just the integer result.

```python
print(5 + 8)
print(6 - 3)
print(2 * 10)
print(8 / 2)
print(13 % 4)
print(8 // 5)
print(2 ** 3)
```

To compare two different values, we use **comparison** operators. These operators check whether the condition holds and returns a boolean for its truth value. The '**==**' operator checks if two values are equal, '**!=**' checks whether they are not equal, '**>**' checks if the first value is greater, and '**<**' checks if the second value is greater. For a broader range of functionality, '**<=**' and '**>=**' checks if the first value is less than or equal and greater than or equal to the second value, respectively.

```python
if 5 < 2:
  print("True")
  print("5 is less than 2")
else:
  print("False")
  print("5 is not less than 2")
print(5<=5)
print(5!=5)
if 0 == 0:
  print("0 is 0")
```

Using **logical** operators, we can combine two conditional statements, like the ones created using comparison operators. The **and** operator returns true if both staments are true and false otherwise, the **or** operator only returns true if at least one statement is true, and the **not** operator returns the opposite boolean value of the conditional statement it is applied to.

```python
if 3 >= 2 and 3 <= 4:
  print("True")
if 0 != 0 or 1 == 1:
  print("Or holds")
print(not True)
```

The **membership** operator '**in**' returns true if the given value is found in the object. Conversely, the '**not in**' operator only returns true if the given value cannot be found in the object.

**Identity** operators check if two objects are the same. The operator '**is**' returns true if the both are the same object and '**is not**' returns true if they are not.

```python
fruits = ["apple", "orange", "pear", "banana", "peach"]
print("apple" in fruits)
print("pineapple" not in fruits)
if type(fruits) is list:
  print("fruits is a list")
if fruits is not fruits:
  print("0 is not an int")
```

**Bitwise** operators are used to perform operations in base 2, using bits. The '**&**' operator compares each bit in two given numbers and sets each result bit to 1 if both bits are also 1 at that position. The '**|**' operator sets each reult bit to 1 if either of the values' bit is 1 at that location. Similarly, the '**^**' operator sets each reult bit to 1 if only one of the values' bit is 1 at that position. The '**~**' operator simply flips all the bits into their oppposite value. The leftward shift operator '**<<**' shifts all the bits to the left by a given amount, filling in the spaces with zeroes. This is equivalent to multiplying the value by a power of two. Meanwhile, the rightward shift '**>>**' shifts the bits to the right and divides the value by some power of two.

```python
print(20 & 15)
print(5 | 4)
print(10 ^ 2)
print(~16)
print(3 << 2)
print(3 * 2**2)
print(8 >> 1)
print(8 / 2)
```

The most basic **assignment** operator is '**=**'. Assignment operators can also combine with arithmetic operators to perform a mathematic operation and assign the value back to a variable. Additionally, bitwise operators can also be combined with the assignment '**=**' to perform a bitwise operation and assign the new value.

```python
count = 1
for i in range(1, 4):
  count *= i
print(count)
count %= 4
print(count)
count **= 4
print(count)
count //= 3
print(count)
count <<= 2
print(count)
count ^= 2
print(count)
```

If the number of parameters passed into the function changes or is unknown, the '**\***' symbol is used to indicate a **tuple** of arguements. The format def func_name(*args): passes a tuple called arg whose elemts can be accessed using indexing and tuple methods.

To pass an unknown number of arguements in the key = value format, as a dictionary, the '**\****' symbol can be used. These are called **keyword arguements**. The header will have the format: def func_name (**kwargs):, and the arguements can be accessed using dictionary methods.

```python
def print_people(*people):
  for person in people:
    print(person)

# as shown below, a varying amount of parameters can be passsed into the function
print_people("Aiden", "Max", "Carly")
print_people("Lily", "Zara", "Ethan", "Frankie", "Rose", "Taylor", "Riley", "Avery")

def print_info(**info):
  for key, value in info.items():
    print(f"{key}: {value}")

# the call passes a varying amount of parameters in key = value format
print_info(me = "Riley", mom = "Susan", dad = "Conner", sister = "Sarah", brother = "Ethan", aunt = "Karen", uncle = "Joe")
print_info(me = "Frankie", mom = "Janet", brother = "Max")
```

**Lambda functions** are small, anonymous functions that use the **lambda** keyword in their definition. As indicated before, they have no name and can only have one expression within their 'body.' These functions are used to simplify a block of code that is only needed once or for one-line functions. They use the format lambda args: expression.

```python
# the lambda function below finds the square of the sum of its parameters
sum_square = lambda *args: sum(args)**2
print(sum_square(1, 2, 3, 4, 5))
```

The **map**() function is an in-built Python function that applies a function with a single arguement and return value to every item in an object. The object must be able to be looped over, or **iterable**. Common iterable objects are lists, sets, tuples, dictionaries, etc. The format of the function call is map(function, iterable)

The **filter**() function also applies a function to every element in an iterable object, but the function must return a boolean value. If the function returns **True**, the item is kept, otherwise it is discarded. The final return value of the filter(function, iterable) call is a version of the iterable object with only the items that returned true.

**reduce**(), unlike the other two in-build functions, returns a single value. It also must be imported using the functools module. This function is applied cumulatively to reduce that iterable to a single meaningful value. The function call is reduce(function, iterable), and the function must take two parameters and return a single value.

We have used the **sorted**() method on the list, set, dictionary, and tuple datatypes. The function can take only one arguement, the iterable type, with the call sorted(iterable). However, to sort by a specific parameter, we can also pass a function to sort by in the key = function format. To sort by something specific, the call sorted(iterable, key = function) is used. We can also pass the 'reverse = True/False' arguement into the call to indicate whether the sort should be ascending or descending; the default is False or ascending.

For all of the functions mentioned in the above methods, we can use lambda to simplify the code. These are the perfect applications for lambda functions since the functions required have only one return value and are simple.

Read and text or **"rt"** is the default mode if not specified. If you want to modify, add to the file, create a new file that does not exist, or view an image, the other modes can be used. The first four modes can be combined with the last to specify the actions that you wish to perform and type of the file.

Another way to access a file, which can reduce possible errors, is using the '**with** keyword. Using this keyword, we can put all file operations inside the body of the header: '**with open("pathname", "mode") as file:**. If opening the file without this keyword and format, we must use the close() method on the file after it has no more use to avoid errors.

While files with the .txt extension can be directly read and modified after being opened, .json, .csv, .pdf, and others may require additional modules and function calls.  

```python
file = open("sales_historical_cars.csv", "wt")
file.close()
with open("sales_historical_cars.csv", "r") as file2:
  print(file2.name)
```

For the all of the lines in a file, the readlines() call will create a list of the lines as strings.

# Object-Oriented Concepts

Python is an **object-oriented** programming language. Thus, everything in Python is an **object**: strings, lists, dictionaries, ints. **Classes** are the blueprints for objects; they define the data contained within the object and the methods that can be used on it. To define a class, we use the class keyword and give it a name. The header will look like class Name:.

Within the body of body of the class, we can define variables and functions that each object created using it will have. Variables in the object are accessed using varname.property. The class's function can be called similarly, with varname.funcname(args). To modify the value kept in a variable in the object, we can use varname.property = new_value.

To create an object using a class, we assign the return value of ClassName(), an instance of the class, to a variable. This variable will be able to acccess the data and methods inside the class automatically. To delete properties from an object or the object itself, we use the **del** keyword.

```python
class Hello:
  greeting = "hello"

# assign an instance of Hello to a variable
hello_object = Hello()

# access and print the data
print(hello_object.greeting)

# we can modify the greeting attribute to "hi"
hello_object.greeting = "hi"
print(hello_object.greeting)

del hello_object
```

The current instance of the class, or the object, is automatically passed into any function call on it as the first parameter. Thus, we must account for this by adding a parameter for the object itself within each function in the class. Usually, this is done with a parameter name '**self**', but it does not matter.

```python
class Math:
  def sum(self, x, y):
    return x + y
  def difference(obj, x, y):
    return x - y
  def product(self, x, y):
    return x * y
  def quotient(name, x, y):
    return x / y

arithmetic = Math()
print(arithmetic.sum(10, 2))
print(arithmetic.difference(9, 3))
print(arithmetic.product(5, 4))
print(arithmetic.quotient(9, 2))
```

All classes have two in-built functions: __init__() and __str__(). The init functions is automatically called when an object of the class is created. This function assigns values to the object's attributes, or the data that change for each instance of the class. These values are passed into the constructor with ClassName(values). Then, in the init function, the properties can be initialized using self.varname = value.

The str() function returns the function's string representation. The default representation is the "<current_module.ClassName object at memory_address". To return a more readable string when printing an object of the class, we must define the __str__() function and specify the string it should return

```python

class Person:
  # while it is good practice to match the attribute names to the parameter names, his is not necessary
  def __init__(self, first_name, last_name, age):
    self.first_name = first_name
    self.last_name = last_name
    self.old = age

person_1 =  Person("Sierra", "James", 29)
print(person_1.first_name)
print(person_1.last_name)
print(person_1.old)
# the class has no defined __str__() function, so it will return the default
print(person_1)

# the Person2 class is identical to Person, except for its __str__() method
class Person2:
  def __init__(self, first_name, last_name, age):
    self.first_name = first_name
    self.last_name = last_name
    self.old = age
  def __str__(self):
    return f"{self.first_name} {self.last_name} is {self.old} years old"

person_2 = Person2("Sierra", "James", 29)
# this will print the specified string representation
print(person_2)
```

To reuse code, we can also have '**child**' classes (or subclasses) which **inherit** functionality from another (**parent**/super) class. To create a child class, we must pass the parent class directly into the child class definition: class ChildName(ParentName):. The child class will automaticallly inherit all methods and data from the parent class.

To refine a method from the parent class, we can write the same method name in the child class to override the original code. Then, the parent class's code can be refrenced using the **super**().funcname() call. Below the call, additional code can be added to make the function more specific. If the parent code is not necessary at all, the super() method does not need to be called.

One child inheriting from a parent is **single inheritance**. If multiple children inherit from a single parent, that would instead be a case of **hierarchical inheritation**.

```python
class Animal:
  def __init__(self, species, name, age, gender):
    self.species = species
    self.name = name
    self.age = age
    self.gender = gender
  def __str__(self):
    return f"{self.name} is a {self.species} and is {self.age} years old"
  def speak(animal):
    return f"**sounds**"

# the child class Human inherits from Animal
class Human(Animal):
  # the init function calls super to initialize the species, first name, age, and gender
  # the unique parameters, last name and location, are initialized individually
  def __init__(self, first_name, last_name, age, country, gender):
    super().__init__("human", first_name, age, gender)
    self.last_name = last_name
    self.location = country
  # the str function adds to the inherited string for a more specific description
  def __str__(self):
    if self.gender == "male":
      return super().__str__() + f". He is from {self.location}"
    else:
      return super().__str__() + f". She lives in {self.location}"
  # the speak function does not use any of the original code
  def speak(person):
    return f"Hello, my name is {person.name}"

class Dog(Animal):
  # the init function calls super to initialize the species, first name, age, and gender
  # the unique parameters, breed and owner, are initialized individually
  def __init__(self, name, age, gender, breed, owner):
    super().__init__("dog", name, age, gender)
    self.breed = breed
    self.owner = owner
  # the str function adds to the inherited string for a more specific description
  def __str__(self):
    if self.gender == "male":
      return super().__str__() + f". He is a {self.breed} and his owner is {self.owner.name}. They live in {self.owner.location}"
    else:
      return super().__str__() + f". She is a {self.breed} and her owner is {self.owner.name}. They live in {self.owner.location}"
  # the speak function does not use any of the original code
  def speak(animal):
    return f"**bark bark**"

# this uses the parent data and functions
animal = Animal("dog", "Max", 5, "male")
print(animal)
print(animal.speak())

# this uses the human class data and functions
human_1 = Human("James", "Carter", 35, "USA", "male")
print(human_1)
print(human_1.speak())

# this uses the dog class data and functions
dog = Dog("Max", 5, "male", "Labrador", human_1)
print(dog)
# despite being the same species as the animal var, dog.speak() returns "*bark bark*"
# since it is a member of the Dog class and overrides the original method
print(dog.speak())
```

Method overriding in child classes is a form of **polymorphism**. This concept describes when a single function or operator has different behaviors based on the way it is used. The two types of polymorphism are **compile-time** and **run-time**.

**Compile-time** polymorphism allows methods or operators with the same name to function differently based its parameters or usage. For example, the **'+'** operator can add two numerical values or concatenate two strings, based on the way it is used.

**Run-time** polymorphism is when objects of different classes respond differently to the same function call. When a child class overrides a method in its parent class, it is decided at runtime that the child class's version of the method should be executed.

```python
def add(x, y):
  return x + y
print(add(1, 2))
print(add("hello", " world"))
```

Just like child class can inherit from a parent class, another class can inherit from it. This describes the concept of **multi-level inheritence**, a chain of classes inheriting from each other. This child class will inherit all data and methods from its parent class and all the information that the parent inherited from the grandparent class.

```python
class Student(Human):
  # the init function calls super to initialize the species, first name, age, and gender
  # the unique parameters, last name and location, are initialized individually
  def __init__(self, first_name, last_name, age, country, gender, student_id, gpa):
    super().__init__(first_name, last_name, age, country, gender)
    self.student_id = student_id
    self.gpa = gpa
  # the grandparent class method is used with super().super() to access the original __string__() method and add to it
  def __string__(self):
    if self.gender == "female":
      return super().super().__str__() + f". She is student {self.student_id} in {self.location} and her gpa is {self.gpa}"
    else:
      return super().super().__str__() + f". He is student {self.student_id} in {self.location} and her gpa is {self.gpa}"
  # this new method adds class-specific functionality
  def is_honor(self):
    if self.gpa > 3.5:
      return True
    else:
      return False

student3 = Student("Cassie", "Daniels", 16, "Candaa", "female", 12345, 4.0)
print(student3)
# even though the speak method is not defined in the Student class, it is inherited from human
print(student3.speak())
if student3.is_honor():
  print("This student is an honor student")
```

**Multiple inheritance** is when a child class inherits from multiple parents, gaining the data and methods of each of them. To inherit from multiple parent classes, they must de specified in the child class definition: class ClassName(Parent1, Parent2):.

Since there are multiple parent classes, super() calls only the first parent class named in the child class definition. Similarly, if there are two functions with the same names in both parent classes, calling the function on an object of the child class will use the first parent's version. This is due to the **Method Resolution Order (MRO)**, in which when a method not in the child class is called, its definition is searched for in the parent classes from left-to-right. The first definitions matching the name that is found is the one that is called. To overrride this preference for the first parent, we can mention another parent by name with the format: Parent2.func_name(child_object).

The **diamond inheritance problem** describes the only situation where the first parent is not automatically preffered. Assume for a class ClassD, its two classes, ClassB and ClassC, inherit a method func_1 from a parent ClassA. If func_1 is overriden in both ClascB and ClassC or only overriden in ClassB, ClassD will inherit ClassB's version of func_1. However, if func_1 is only overriden in ClassC, ClassD will inherit ClassC's version of the function. This is an example of hybrid inheritance, since it mixes different inheritance types: multi-level and multiple.

```python
class Parent1:
  def func_1(self):
    print("Parent 1")
  def func_2(self):
    print("Parent 1")
class Parent2:
  def func_1(self):
    print("Parent 2")
  def func_2(self):
    print("Parent 2")
class Child(Parent1, Parent2):
  # Parent2's version of func_2 is specfically called
  def func_2(self):
    Parent2.func_2(self)

obj = Child()
# since Parent1 is first in the class definition, it's version of func_1 is inherited
obj.func_1()
# Parent2's version of func_2 is used in the Child class override
obj.func_2()


class ClassA:
  def func_1(self):
    print("Class A")
  def func_2(self):
    print("Class A")
  def func_3(self):
    print("Class A")
  def func_4(self):
    print("Class A")
class ClassB(ClassA):
  def func_1(self):
    print("Class B")
  def func_2(self):
    print("Class B")
class ClassC(ClassA):
  def func_1(self):
    print("Class C")
  def func_3(self):
    print("Class C")
class ClassD(ClassB, ClassC):
  pass

obj = ClassD()
# since both classes override func_1, ClassB's version is printed
obj.func_1()
# since ClassB overrides func_2, its version is printed
obj.func_2()
# since ClassC overrides func_3, its version is printed
obj.func_3()
# since neither class overrides func_4, ClassA's version is printed
obj.func_4()
```

One of the core principles of object-oriented programming is **abstraction**. This describes the concept of hiding internal details and only allows necessary features to be shown. Abstraction can be completed using **abstract classes** and **abstract methods**. Abstract classes can only be base classes in Python, meaning they cannot inherit from other user-created classes. They also cannot be instantiated. Abstract methods, however, must be implemented/overriden in their child classes.

To create an abstract class, we must first import **'ABC'** and **'abstractmethod'** from the **abc** module. Any abstract method must have ABC as its only parent class. Within the body of the class, regular functions and abstract functions can both be defined. Normal functions will be inherited normally by children, but abstract methods must be overriden. Abstract methods are indicated with '**@abstractmethod** before the function definition. Within the abstract method body, there should only be the keyword **'pass'** to leave implementation for child classes.

If every method within an abstract class is also abstract, then it is **fully abstract**. Otherwise, there is only **partial abstraction**.

```python
from abc import ABC,abstractmethod

# this class is partially abstracted since it contains non-abstract methods
class Wings(ABC):
  @abstractmethod
  def canFly(self):
    pass
  @abstractmethod
  def canGlide(self):
    pass
  @abstractmethod
  def canSwim(self):
    pass
  def __str__(self):
    return f"This creature has wings."

# all abstract methods must be implemented in subclasses, regular functions can be overriden if wished
class FruitFly(Wings):
  def canFly(self):
    return True
  def canGlide(self):
    return True
  def canSwim(self):
    return True
  def __str__(self):
    return super().__str__() + f" It can fly, glide, and swim."

class SugarGlider(Wings):
  def canFly(self):
    return False
  def canGlide(self):
    return True
  def canSwim(self):
    return False
  def __str__(self):
    return f"This animal has membranes that mimic wings and can glide"

class Penguin(Wings):
  def canFly(self):
    return False
  def canGlide(self):
    return False
  def canSwim(self):
    return True
  def __str__(self):
    return super().__str__() + f" It can only swim."

fruitfly = FruitFly()
print(fruitfly)
if fruitfly.canFly():
  print("It can fly")
if fruitfly.canGlide():
  print("It can glide")
if fruitfly.canSwim():
  print("It can swim")

sugar = SugarGlider()
print(sugar)
if sugar.canFly():
  print("It can fly")
if sugar.canGlide():
  print("It can glide")
if sugar.canSwim():
  print("It can swim")

penguin = Penguin()
print(penguin)
if penguin.canFly():
  print("It can fly")
if penguin.canGlide():
  print("It can glide")
if penguin.canSwim():
  print("It can swim")
```

Another important OOP concept is **encapsulation**, where data and the methods that use them are bundled into a single object: a class. Additionally, the access to this data is restricted to prevent it from being modified without necessary permission or misused and reduce complexity for the user.

**Public** attributes are defined as normal variables and are accessible everywhere. They can be directly accessed and modified using object.property = value.

By putting and underscore **'_'** before an attribute's name, we can make it a **protected** member. This data should only be accessed and modified within the class and its subclasses. While the variables can still be accessed and modified anywhere, the underscore is meant to discourage this.

The highest level of encapsulation is **private** data, which can only be accessed within the class itself. Members are made private by putting two underscores **'__'** before the property name. When these attributes are accessed, they raise an AttributeError. This is because they are renamed to _ClassName__var_name. The variable can still be modified using this new name, but it is meant to prevent this. To view and change the variable safely, the class should include getters and setters in the methods.

```python
class BankAccount():
  def __init__(self, name, balance):
    self.name = name
    # the balance is private, because it should not be modified without permission
    self.__balance = balance
  def deposit(self, amount):
    self.__balance += amount
  def withdraw(self, amount):
    self.__balance -= amount
  def get_balance(self):
    return self.__balance

account = BankAccount("Sarah Palin", 50)

# accessing the private variable directly will raise an error
try:
  account.__balance +=5
except AttributeError:
  account._BankAccount__balance += 5

print(f"balance: {account.get_balance()}")
```
