**Data Preprocessing**

```python
import pandas as pd
df = pd.read_csv("sales_historical_cars.csv")
df

```

```python
df = df[df['parameter'] == 'EV sales']
df = df[df['region'] != 'World']
df = df[df['region'] != 'Europe']
df = df[df['region'] != 'EU27']
df = df[df['region'] != 'Rest of the world']
df.drop(columns = ['category', 'mode', 'unit'], inplace = True)
df
```

**Investigating Representation of Countries in EV Sales**


---



```python
import matplotlib.pyplot as plt
import numpy as np
counts = []
for country in df['region'].unique():
  counts.append(df[df['region'] == country]['value'].sum())
plt.figure(figsize=(10, 10))
plt.barh(df['region'].unique(), counts, color = 'violet')
plt.title('Frequency of Countries')
plt.xlabel('EV Car Sales (10M)')
plt.ylabel('Country')
plt.show()
```

As expected, the country with one of the highest populations has the highest number of car sales. Surprisingly, the country with the largest population, India, has some of the lowest sales. Overall, East Asian and European countries lead in EV sales, in addition to the U.S. and Australia.
